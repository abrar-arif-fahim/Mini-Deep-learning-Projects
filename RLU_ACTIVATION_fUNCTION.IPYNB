```python
import numpy as np
from Sigmoid_fnc_scratch import sigmoid
import matplotlib.pyplot as plt
def plot_graph(fn, x=None, **kwargs):
    if x is None:
        x = np.linspace(-10, 10, 400)
    plt.plot(x, fn(x), **kwargs)
    plt.xlabel("x")
    plt.ylabel("y")
    plt.grid(True)

# example usage
plot_graph(sigmoid)
```
